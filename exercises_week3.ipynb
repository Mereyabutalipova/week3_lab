## 7. Manipulating Tensors (Basic Operations)

### Exercise 7.1: Perform basic operations
Create a tensor [10, 20, 30]. Add 5, subtract 10, multiply by 2, and divide by 10.

```python
tensor = torch.tensor([10, 20, 30])
# TODO: Add 5
tensor_add = tensor + 5

# TODO: Subtract 10
tensor_sub = tensor - 10

# TODO: Multiply by 2
tensor_mul = tensor * 2

# TODO: Divide by 10
tensor_div = tensor / 10

print(f"Original: {tensor}")
print(f"Addition (+5): {tensor_add}")
print(f"Subtraction (-10): {tensor_sub}")
print(f"Multiplication (*2): {tensor_mul}")
print(f"Division (/10): {tensor_div}")
```

### Exercise 7.2: Matrix multiplication
Create two tensors A (2x3) and B (3x2) with random values and perform matrix multiplication.

```python
# TODO: Create A and B
A = torch.rand(2, 3)
B = torch.rand(3, 2)

# TODO: Matrix mul using torch.mm
result_mm = torch.mm(A, B)

# TODO: Matrix mul using @
result_at = A @ B

print(f"Matrix A:\n{A}")
print(f"Matrix B:\n{B}")
print(f"\nResult (using torch.mm):\n{result_mm}")
print(f"Result (using @):\n{result_at}")
print(f"\nResult Shape: {result_at.shape}")
```

## 8. Tensor Aggregation

### Exercise 8.1: Find min, max, mean, sum
Create a tensor with values from 0 to 100 and find its min, max, mean, and sum.

```python
agg_tensor = torch.arange(0, 101)
# TODO: Min
tensor_min = agg_tensor.min()

# TODO: Max
tensor_max = agg_tensor.max()

# TODO: Mean (note: may need to cast to float)
tensor_mean = agg_tensor.type(torch.float32).mean()

# TODO: Sum
tensor_sum = agg_tensor.sum()

print(f"Minimum: {tensor_min}")
print(f"Maximum: {tensor_max}")
print(f"Mean: {tensor_mean}")
print(f"Sum: {tensor_sum}")
```

### Exercise 8.2: Argmin and argmax

```python
agg_tensor = torch.arange(0, 101)

# TODO: Argmin
min_index = agg_tensor.argmin()

# TODO: Argmax
max_index = agg_tensor.argmax()

print(f"Index of minimum: {min_index}")
print(f"Index of maximum: {max_index}")
```
